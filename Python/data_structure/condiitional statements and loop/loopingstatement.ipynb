### looping statements are used for iterartions

range()functions cann be used for iterations

## range() takes an input parameter and the first parameter that the range will be taking as the input will be the initial value that you want to priht or you want to operate on and then you will be passing the next input after comma as the parameter as the final value and it can take third parameter as well and that will be the   step size and that is the difference between the iterating values

```python
# print the numbers from 1 to 50
#range(initial_value, final_value+1,step_size)
for item in range(1,51,3):
    print(item)
```

```python
#no from 0 to 100
for i in range(100):
    print(i)
```

```python
for i in range(100,0,-1):
    print(i)
```

## while loop is beilg used while iterating when conditions are true then only the code is printed

```python
# print 0 to 50 using while loop
i=0
while i<=50:
    print(i)
    i+=1
```

```python
# print even no netwwen 0 to 50 using while loop
i=0
while i<=50:
    if i%2==0:
        print(i)
    else:
        pass #pass is used to do control the statemts in python
    i+=1
```

### break is a loop control statement

```python
for i in range(1,51):
    if i==18:
        break
    else: 
        print(i)
```

### continue statemt is used to contol the loop statement while skiiping the values untill the condions are met

```python
for i in range(1,51):
    if i in range(18):
        continue
    else:
        print(i)

```

### f string--> f string will be helping us out with printing the numbers with the variable value present inside the colons so we can directly represent the variable value inside the colons message itself where variable value will be shown inside the {}

```python
# Nested loops can help us out with iterating over a series of numbers as with the
# proper combination of iterating numbers
for i in range(6):
    for j in range(4):
        print(f"i value is: {i} and j value is: {j}")
```

```python
## printing the sum of First N natural numbers where Natural Number--->1,2,3....n
n_value=int(input("Enter the value for n :"))
sum=0
for i in range(n_value+1):
    sum+=i
print(sum)
print(f"The sum of first {n_value} natural numbers is : {sum}")
```
