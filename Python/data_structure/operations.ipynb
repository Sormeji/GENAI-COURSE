```python
#Simple Calculator
a=int(input("Enter the first value: "))
b=int(input("Enter the second value:"))
operation=input("Enter the operation out of +, -, *, /")

if operation == "+":
    print("The addition of the two values is:", a+b)
if operation == "-":
    print("The substraction of the two values is:", a-b)
if operation == "*":
    print("The product of the two values is", a*b)
if operation == "/":
    print("The division of the two values is:", a/b)
```
