### C onditional statements working like a real life situation if a codition in meeeting the the code get executed

```python
# Eligibiity for driving on meeting a certain condition
## Conditional Statements---> if statements where if a condition is meeting
# then a block of code gets executed
Age=int(input("Enter your age value please :"))
if Age>=18:
    print("Yes you are eligible for driving")
else:
    print("You are not eligible for driving")
```

```python
## Figure out that who is the tallest among the three friends
Amit_height=int(input("Enter that what your height is Amit:"))
Prabhat_height=int(input("Enter that what your height is Prabhat: "))
Shivy_height=int(input("Enter your height value Shivy :"))
if Amit_height>Prabhat_height>Shivy_height:
    print("Amit is the tallest among the three friends")
elif Prabhat_height>Amit_height>Shivy_height:
    print("Prabhat is the tallest among the three friends")
else:
    print("Shivy is the tallest among the three friends")
```
