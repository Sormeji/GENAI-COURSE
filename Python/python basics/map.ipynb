## MAP FUNCTION

## Map function applies a given function operation to All items in an iterable or a List and returns a Map Object (An Iterator)

## Map function is particularly useful for Transforming the data or applying an operation on ALL items of the iterable
## Map() is going to help us out with avoiding unnecessary lines of code and we can apply the operation on all values or all items in an iterable and this is the sign of a good programmer 

## Syntax for map
### list(map(function,iterable))

```python
## Squaring all items of a list
## list---> [1,4,5,3,8,6,9]
## []--->[1,16,25,,,81]....
def square(list_items):
    empty_list=[]
    for item in list_items:
        squared_item=item**2
        empty_list.append(squared_item)
    print(empty_list)
```

```python
square([1,4,5,3,8,6,9])
```

```python
## similar problem statement with a Map()--->
print(list(map(lambda x:x**2, [1,4,5,3,8,6,9])))
```

## Whenever you are going to pass the general function inside your map function as the parameter then you dont have to use parenthesis

## While Mapping the value of one iterable with another iterable we can actually apply the operation in between the values of the two iterables in a synchronous way
#### [1,2,3,4] <---> [5,6,8,2]---> Addition ---> [6,8,11,6]


```python
## Map Function with Multiple Iterables
numbers_1=[1,4,5,8]
numbers_2=[8,5,6,7]
print(list(map(lambda x,y:x+y, numbers_1, numbers_2)))
```

```python
### Map() To convert all string values items in an iterable into an integer
print(list(map(int, ['1', '2', '9', '5', '6', '8'])))
```

```python
employees_name=['amit','juhi', 'shivy', 'kunal', 'rohan', 'abhishek']
print(list(map(str.upper, employees_name)))
```

```python
### Retrieve the values inside a dictionary using Map()
employees=[{'Name': 'Amit', 'Age':23, 'Qualification': 'Btech'},
{'Name': 'Vivek', 'Age': 27, 'Qualification': 'BBA'},
{'Name': 'Shivy', 'Age': 26, 'Qualification': 'Bcom'}]
```

```python
def age_retiever(list_item):
    return list_item['Age']

```

```python
print(list(map(age_retiever, employees)))   
```
