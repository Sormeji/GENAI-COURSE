```python
# Create a function that converts temperature from celcius to Fahrenheit or
# Fahrenheit to Celcius
## if temperature is in celcius then celcius to fahrenheit conversion is:
'''1 deg celcius = 9/5 + 32
45 deg C=45^ * 9/5 + 32 F
1 Fahrenheit celcius =5/9 -32
45 deg F = (45-32) *5/9 deg C '''
def temperature_change (temp_value, unit):
    if unit=='C':
        temperature=temp_value *9/5 +32
        print (f"The temperature for {temp_value} deg celcius is Fahrenheit is:", str(temperature)+" deg F")
    elif unit=='F':
        temperature=round((temp_value-32) *5/9,2)
        print (f"The temperature for {temp_value} deg fahrenheit is Celsius is:", str(temperature)+" deg C")
    else:
        print("Please enter the correct unit of temperature")
```

```python
temperature_change(45, 'C')
```

```python
temperature_change(45, 'F')
```

```python
## define a Function that would behave as a Password Strength Checker
''''Strong password-->
if your password:
1. length is greater than 8 characters
2. password must consist of atleast a lowercase character (a,b,c...z)
3. password must consist of atleast a uppercase character (A,B,C...Z)
4. password must consists of atleast a digit (1,2,)
5. password must consists of atleast a special character (!, @, #, $,^,&,*,(,),_) If a password matches all above conditions then it will be considered as a strong password
Weak passwords-->
If a password doesnt match any above condition then it will be considered as a weak password'''
```

```python
def password_strength(password):
    if len(password) < 8:
        return "Weak password: Password must be at least 8 characters long."
    elif not any(char.islower() for char in password):
        return "Weak password: Password must contain at least one lowercase letter."
    elif not any(char.isupper() for char in password):
        return "Weak password: Password must contain at least one uppercase letter."
    elif not any(char.isdigit() for char in password):
        return "Weak password: Password must contain at least one digit."
    elif not any(char in "!@#$%^&*()_+" for char in password):
        return "Weak password: Password must contain at least one special character (!@#$%^&*()_+)."
    else:
        return "Strong password: Your password meets all the criteria."    
```

```python
password_strength("Satyajit@123")
```

```python
## Check that whether if the current string is a palindrome or not
def palindrome (string):
    string_value=string.lower().replace(" ", "")
    if string_value==string_value[::-1]:
        print("The current string is a palindrome")
    else:
        print("The current string is not a palindrome")
        
```

```python
palindrome("Aabb aa")
```

```python
palindrome("A man a plan a canal Panama")
```
