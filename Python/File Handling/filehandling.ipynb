File Handling
1. Read The content from My Text File in the system

```python
## open()--> opening your file
## First Parameter --> File_name
## Second Parameter --> Mode
## Mode --> It decides the way in which we are working with this file
'''Mode --> --> Reading the content from the File
Mode --> w --> Write the content into the File
Mode
---> r+ -->
Read the content first and then writing the content into the File
Mode ---> w+ ---> Writing the content first and then reading from it'''
```

```python
## Reading your File
## read()--> Reads the content from the File
```

```python
f=open('example.txt','r')
content=f.read()
print(content)
f.close()
```

## Writing the Content into the File
## write()--> it writes the content into the text file
## Mode--> w --> Content will be overwritten
## Mode ---> a ---> Content will be added from the end

```python
## write()--> helps us out in writing the content into the File
```

```python
f=open('example.txt', 'w')
f.write('Hello We are now writing the content into our File\n')
f.close()
```

```python
## Appending the content into our File
f=open('example.txt','a')
f.write('We are learning about Gen AI Pre requisite Python')
f.close()
```

```python
## with --> It will be used for first of all opening the file as a object and there is no requirement for closing the file as well
```

```python
with open('example.txt','r') as f:
    content_data=f.read()
    print(content_data)
```

```python
with open('example.txt', 'w') as file:
    file.write("We are exploring about with command in python File Handling")
```

```python
with open('example.txt', 'a') as files:

    files.write("\nHello we are exploring append mode as well")
```

```python
## Write the content into our file line by line--> writelines()

lines=['First_Line\n', 'Second Line\n', 'Third Line\n']
with open('example.txt', 'w') as data:
    data.writelines (lines)
```

```python
### Read the content from the example.txt file and then writing the content into another file

with open('source.txt','r') as source:
    content=source.read()
with open('destination.txt', 'w') as destination:
    destination.write(content)
```

```python
## Reading the Total Lines, Total Words Total Characters from a file-->readlines()
def count_lines(file):
    with open(file, 'r') as files:
        lines = files.readlines()
    total_lines = len(lines)
    total_words = sum(len(line.split()) for line in lines)
    total_characters = sum(len(line) for line in lines)
    return total_lines, total_words, total_characters

file_path='Total_data.txt'
Total_Lines, Total_Words, Total_Characters=count_lines(file_path)
print (f"Total Lines are: {Total_Lines} and Total Words are: {Total_Words} and Total Characters are: {Total_Characters}")
```
